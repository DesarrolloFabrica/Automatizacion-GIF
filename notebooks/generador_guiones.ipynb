# Generador de guiones académicos

Flujo para cargar un sílabo, revisar los cinco temas detectados y generar los documentos base en Word.

## 1. Instalación

Ejecuta esta celda una vez si el entorno no tiene las dependencias.

```python
import sys
import subprocess

print('Python del kernel:', sys.executable)
subprocess.check_call([
    sys.executable,
    '-m',
    'pip',
    'install',
    'python-docx',
    'pypdf',
    'openai',
    'python-dotenv',
])
```

## 2. Configuración

La API key debe estar en la variable de entorno `OPENAI_API_KEY`. Evita escribirla directamente en el notebook.

```python
import os
from pathlib import Path

BASE_DIR = Path('..').resolve()
SYLLABUS = BASE_DIR / '4 Syllabus Fisiopatología Humana.docx'
OUTPUT_DIR = BASE_DIR / 'outputs'
MODEL = os.getenv('OPENAI_MODEL', 'gpt-4o')

# En PowerShell, configura antes de abrir Jupyter:
# $env:OPENAI_API_KEY="TU_API_KEY"
```

## 3. Revisar plan detectado sin consumir API

```python
import sys
sys.path.append(str(BASE_DIR))

print('BASE_DIR:', BASE_DIR)
print('SYLLABUS existe:', SYLLABUS.exists(), SYLLABUS)

from generate_guiones import extract_course_plan

plan = extract_course_plan(SYLLABUS)
plan
```

## 4. Generar documentos

Esta celda llama a la API y puede tardar varios minutos porque genera cinco documentos largos.

```python
import subprocess

cmd = [
    sys.executable,
    str(BASE_DIR / 'generate_guiones.py'),
    '--syllabus', str(SYLLABUS),
    '--output-dir', str(OUTPUT_DIR),
    '--model', MODEL,
]

subprocess.run(cmd, check=True)
```
